# **MANEJO DE ERRORES: EXCEPCIONES**

***
### **EXCEPCIONES**
***

Generalmente cuando durante la ejecución de un programa se encuentra un error, se frena esa dicha ejecución y se imprime el error por pantalla. Las excepciones son bloques de código que permiten continuar la ejecución del código aún cuando se pesente algún error.

Éstas se componen de un término `try` que contiene el bloque de código propenso a generar errores de cualquier tipo y un término `except` que contiene el bloque de código a ejecutar en caso se hallar un error.

La sintaxis es la siguiente:


```python
try:
  bloque de código propenso a tener errores
except:
  bloque de código a ejecutar en caso de hallar un error
```



````python
try:
  numerador = int(input("Ingresa el numerador: "))
  divisor = int(input("Ingresa el divisor: "))
  division = numerador/divisor
  print(division)
except:
  print('Por favor, intente de nuevo.')
````

Para no tener que volver a ejecutar el código cada vez que aparece un error, podemos encerrarlo en un bucle while.

De este modo, cada vez que encuentre errores el código volverá a ejecutarse automáticamente

````python
while True:
    try:
        numerador = int(input("Ingresa el numerador: "))
        divisor = int(input("Ingresa el divisor: "))
        division = numerador/divisor
        print(division)
        break
    except:
        print('Por favor, intente de nuevo.')
````

**Else**


Sirve para especificar un bloque de código que se quiera ejecutar cuando NO se encuentre ningún error.

````python
while True:
    try:
        numerador = int(input("Ingresa el numerador: "))
        divisor = int(input("Ingresa el divisor: "))
        division = numerador/divisor
        print(division)
    except:
        print('Por favor, intente de nuevo.')
    else:
        print('La ejecución fue exitosa.')
        break
````

**Finally**

Sirve para especificar un bloque de código que se ejecute SIEMPRE, sin importar si se encuentra o no algún error.

````python
while (True):
    try:
        numerador = int(input("Ingresa el numerador: "))
        divisor = int(input("Ingresa el divisor: "))
        division = numerador/divisor
        print(division)
    except:
        print('Por favor, intente de nuevo.')
    else:
        print('La ejecución fue exitosa.')
        break
    finally:
        print("Fin del bucle.")
````

**Imprimir el error del programa**

Para detectar el tipo de error es posible asignar la excepción a una variable. Esto permite analizar el tipo de error que sucede.

````python
while (True):
  try:
      numerador = int(input("Ingresa el numerador: "))
      divisor = int(input("Ingresa el divisor: "))
      division = numerador/divisor
      print(division)
      break
  except Exception as e:
      print("Ocurrió el siguiente error: ", type(e).__name__)
````

***
### **EXCEPCIONES MÚLTIPLES**
***

En el mismo bloque de código pueden surgir distintos tipos de errores. Podemos usar un bloque try-except para manejar cada una de las exepciones de manera distinta.

La sintaxis de las excepciones múltiples es la siguiente:

```python
try:
  bloque de código propenso a tener errores
except tipo_de_error_1:
  bloque de código a ejecutar en caso de hallar un tipo_de_error_1
except tipo_de_error_2:
  bloque de código a ejecutar en caso de hallar un tipo_de_error_2
except tipo_de_error_3:
  bloque de código a ejecutar en caso de hallar un tipo_de_error_3
...
```

````python
while (True):
  try:
      numerador = int(input("Ingresa el numerador: "))
      divisor = int(input("Ingresa el divisor: "))
      division = numerador/divisor
      print(division)
      break
  except ValueError:
      print("No se puede dividir un texto. Intenta de nuevo.")
  except ZeroDivisionError:
      print("No se puede dividir por cero. Intenta de nuevo.")
````

Es posible, dentro de las excepciones múltiples, incluir la detección del error `except Exception as e:` para aquellos casos donde el programa falla, pero la falla no se debe a ninguno de los errores listados en las excepciones anteriores.

````python
while (True):
  try:
      numerador = int(input("Ingresa el numerador: "))
      divisor = int(input("Ingresa el divisor: "))
      division = numerador/divisor
      print(division)
      break
  except ValueError:
      print("No se puede dividir un texto. Intenta de nuevo.")
  except ZeroDivisionError:
      print("No se puede dividir por cero. Intenta de nuevo.")
# Esta última línea se incluye por si surge un error no contemplado
  except Exception as e:
      print("Ocurrió el siguiente error: ", type(e).__name__ )
````
